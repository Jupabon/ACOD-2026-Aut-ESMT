## Selecting Data

Selecting data is one of the most common pandas operations.

We will focus on three common situations:

1. Selecting columns.
2. Selecting rows by position or label.
3. Selecting rows using conditions.

We will practise using the wine reviews DataFrame.

### 1. Selecting columns

A single column can be selected using square brackets.

Multiple columns can be selected by passing a list of column names.

### Exercise

```python
wine_subset = reviews[['country', 'province', 'points', 'price']]
wine_subset.head(10)
```

### 2. Selecting rows with `loc` and `iloc`

`loc[]` selects using **labels**.

`iloc[]` selects using **positions**.

```python
# Select rows with index labels 10, 750 and 1200
reviews.loc[[10, 750, 1200]]
```

```python
# Select the first 5 rows and first 4 columns by position
reviews.iloc[:5, :4]
```

<div class="alert alert-warning">
<b>Remember.</b>
<code>loc</code> works with labels and conditions.  
<code>iloc</code> works with numerical positions.
</div>

### Exercise

Create a sub-DataFrame containing the `country`, `province`, `region_1` and `region_2` columns for rows with index labels `10`, `750` and `1200`.

### 3. Selecting rows using conditions

Boolean conditions allow us to keep only observations that satisfy a rule.

```python
reviews.loc[reviews['country']=='France']
```

We can combine conditions using:

* `&` for **and**
* `|` for **or**

Each condition should be placed inside parentheses.

```python
reviews.loc[
    (reviews['country']=='France') &
        (reviews['points']>=95)]
```

`.isin()` is useful when we want to match several possible values.

```python
reviews.loc[
    reviews['country'].isin(['Italy', 'France']) 
]
```

## Inspecting Data Quality

Before analysing a dataset, we should check whether the data look reliable.

Useful questions include:

* Are values missing?
* Are there duplicated rows?
* Do numerical values have plausible ranges?
* How many unique values does each variable contain?
* Are the data types appropriate?

These checks do not automatically tell us what to do, but they help us identify potential problems.

```python
# How many missing values are in each column?
reviews.isnull().sum().sort_values
```

```python
# What proportion of each column is missing?
(reviews.isnull().mean()*100).sort_values(ascending=False)
```

```python
# How many completely duplicated rows are there?
reviews.duplicated().sum()
```

```python
# How many unique values are in each column?
reviews.nunique().sort_values()
```

```python
# Quick numerical summary
reviews.describe()
```

<div class="alert alert-warning">
<b>Important.</b>
Finding missing values does not automatically mean that we should drop them or replace them.

Before changing the data, ask:

<ul>
<li>Why might the value be missing?</li>
<li>How much data would we lose by dropping rows?</li>
<li>Would replacing the value introduce unrealistic information?</li>
<li>Could the missingness itself be informative?</li>
</ul>
</div>

## Mini Challenge

You have been asked to take a first look at **highly rated wines**.

Create a DataFrame containing wines with at least **95 points** and investigate it.

Using pandas:

1. Create a DataFrame called `high_rated` containing wines with at least 95 points.
2. Determine how many wines meet this condition.
3. Determine which countries are represented.
4. Check how many of these wines have a missing price.
5. Calculate the mean and median price.
6. Display the 10 most expensive highly rated wines using only:
   * `country`
   * `title`
   * `points`
   * `price`

As you work, think about what else you might want to know before drawing conclusions from this subset.

```python
#1. Create a DataFrame called `high_rated` containing wines with at least 95 points.
high_rated = reviews.loc[reviews['points'] >= 95]
```

```python
#2. Determine how many wines meet this condition.
high_rated_count = high_rated.shape[0]
print(f'The number of wines with at least 95 points is: {high_rated.shape[0]}')
```

```python
#3. Determine which countries are represented.
high_rated_countries = high_rated['country'].unique()
print(f'The countries represented in the high rated wines are: {high_rated_countries}')
```

```python
#4. Check how many of these wines have a missing price.
high_rated_missing_price_count = high_rated['price'].isnull().sum()
print(f'The number of high rated wines with a missing price is: {high_rated_missing_price_count}')
```

```python
#5. Calculate the mean and median price.
high_rated_mean_price = high_rated['price'].mean()
high_rated_median_price = high_rated['price'].median()
print(f'The mean price of high rated wines is: {high_rated_mean_price}')
print(f'The median price of high rated wines is: {high_rated_median_price}')
```

```python
'''6. Display the 10 most expensive highly rated wines using only:
   * `country`
   * `title`
   * `points`
   * `price`'''
high_rated_expensive = high_rated.nlargest(10, 'price')[['country', 'title', 'points', 'price']]
display(high_rated_expensive)
```
